```python
class Alarma:
    def __init__(self, tipo, estado, ubicacion):
        self.tipo = tipo
        self.estado = estado
        self.ubicacion = ubicacion

    def leerDatos(self):
        self.tipo = input("Ingrese el tipo de alarma: ")
        self.estado = input("Ingrese el estado (Activa/Inactiva): ")
        self.ubicacion = input("Ingrese la ubicación: ")

    def mostrarDatos(self):
        print("Tipo: ", self.tipo)
        print("Estado: ", self.estado)
        print("Ubicación: ", self.ubicacion)

# Creando los objetos alarma1 y alarma2
alarma1 = Alarma("", "", "")
alarma2 = Alarma("", "", "")

print("Alarma Numero 1")
alarma1.leerDatos()
alarma1.mostrarDatos()
print("-*-*-*-*-*-*-*-*-*-*-*")

print("Alarma Numero 2")
alarma2.leerDatos()
alarma2.mostrarDatos()
print("-*-*-*-*-*-*-*-*-*-*-*")

print("Software desarrollado por Nicole Robles NL = 46")
```
